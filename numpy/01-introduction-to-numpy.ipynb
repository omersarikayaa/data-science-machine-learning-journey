## 1. Setup

NumPy is conventionally imported with the alias `np`. Anaconda usually includes NumPy, so a separate installation is generally unnecessary.

```python
import numpy as np

print(f"NumPy version: {np.__version__}")
```

## 2. From Python lists to NumPy arrays

A Python list is a general-purpose container. A NumPy array is designed for efficient numerical computation and supports vectorized operations.

```python
my_list = [10, 20, 30, 40]
my_array = np.array(my_list)

print("List type:", type(my_list))
print("Array type:", type(my_array))
print("Array:", my_array)
```

Useful array attributes:

- `ndim`: number of dimensions
- `shape`: size along each dimension
- `size`: total number of elements
- `dtype`: data type stored in the array

```python
my_numpy_array = np.array([10, 20, 30, 40, 50])

print("Dimensions:", my_numpy_array.ndim)
print("Shape:", my_numpy_array.shape)
print("Size:", my_numpy_array.size)
print("Data type:", my_numpy_array.dtype)
print("Maximum value:", my_numpy_array.max())
```

## 3. Creating arrays

NumPy provides helper functions for creating arrays with predefined or random values.

```python
ones_array = np.ones(5)
zeros_array = np.zeros(5)

print("Ones:", ones_array)
print("Zeros:", zeros_array)
```

## 4. Array arithmetic

The `+` operator concatenates Python lists. With NumPy arrays, arithmetic operators work **element by element**.

```python
list_1 = [1, 2]
list_2 = [2, 3]

array_1 = np.array(list_1)
array_2 = np.array(list_2)

print("List concatenation:", list_1 + list_2)
print("Array addition:", array_1 + array_2)
print("Array subtraction:", array_1 - array_2)
print("Array multiplication:", array_1 * array_2)
print("Array division:", array_1 / array_2)
print("Scalar multiplication:", array_1 * 5)
```

> **Key idea:** Vectorized operations let us apply a calculation to an entire array without writing a Python loop.

## 5. Basic aggregation

Aggregation methods summarize all values in an array.

```python
values = np.array([10, 20, 30, 40, 50])

print("Minimum:", values.min())
print("Maximum:", values.max())
print("Sum:", values.sum())
print("Mean:", values.mean())
```

## 6. Creating sequences with `arange`

`np.arange(start, stop, step)` works similarly to Python's `range`, but returns a NumPy array. The `stop` value is excluded.

```python
python_sequence = list(range(0, 10))
numpy_sequence = np.arange(0, 10)
even_numbers = np.arange(0, 20, 2)

print("Python range as a list:", python_sequence)
print("NumPy sequence:", numpy_sequence)
print("Even numbers:", even_numbers)
```

## 7. Indexing and slicing

Indexing selects a single value, while slicing selects a range of values.

Slice syntax: `array[start:stop:step]`. The `stop` index is excluded, and negative indices count from the end.

```python
numbers = np.arange(0, 10)

print("Original array:", numbers)
print("First element:", numbers[0])
print("Last element:", numbers[-1])
print("Elements from index 1 to 3:", numbers[1:4])
print("Every second element from index 2 to 5:", numbers[2:6:2])
print("Reversed array:", numbers[::-1])
```

## 8. Random number generation

`default_rng` is NumPy's modern random-number interface. A fixed seed makes the results reproducible, which is useful for learning, debugging, and machine learning experiments.

```python
rng = np.random.default_rng(seed=42)

uniform_values = rng.random(5)
normal_values = rng.standard_normal(4)
random_matrix = rng.standard_normal((4, 4))
random_integers = rng.integers(1, 300, size=5)

print("Uniform values between 0 and 1:", uniform_values)
print("Normally distributed values:", normal_values)
print("Random 4×4 matrix:\n", random_matrix)
print("Random integers:", random_integers)
```
