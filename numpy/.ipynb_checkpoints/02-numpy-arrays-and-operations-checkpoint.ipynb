# NumPy Arrays and Operations

This notebook continues my NumPy studies with multidimensional arrays, array creation, views and copies, boolean filtering, and vectorized mathematical operations.

## Learning objectives

- Create and inspect multidimensional arrays
- Generate evenly spaced values with `linspace`
- Understand the difference between a view and a copy
- Filter arrays with boolean conditions
- Apply vectorized arithmetic and universal functions

## 1. Setup

NumPy is conventionally imported with the alias `np`. A seeded random-number generator makes the examples reproducible.

```python
import numpy as np

rng = np.random.default_rng(seed=42)

print(f"NumPy version: {np.__version__}")
```

## 2. Multidimensional arrays

A nested Python list can be converted into a two-dimensional NumPy array. For a matrix, `shape` returns `(rows, columns)`.

```python
matrix_list = [
    [1, 0, 0],
    [0, 1, 0],
    [0, 0, 1],
    [0, 0, 0],
]

matrix = np.array(matrix_list)

print("Matrix:\n", matrix)
print("Dimensions:", matrix.ndim)
print("Shape:", matrix.shape)
print("Total number of elements:", matrix.size)
```

Values can be selected with `array[row, column]`. Slices can select complete rows, columns, or rectangular sections.

```python
print("First element:", matrix[0, 0])
print("Second row:", matrix[1])
print("Second column:", matrix[:, 1])
print("Top-left 2×2 section:\n", matrix[:2, :2])
```

## 3. Creating arrays with a specific shape

`zeros` and `ones` accept a tuple that describes the desired shape. They create floating-point arrays by default, but `dtype` can be specified explicitly.

```python
zero_matrix = np.zeros((3, 4), dtype=int)
one_matrix = np.ones((2, 5))
identity_matrix = np.eye(3, dtype=int)

print("3×4 zero matrix:\n", zero_matrix)
print("2×5 one matrix:\n", one_matrix)
print("3×3 identity matrix:\n", identity_matrix)
```

## 4. Evenly spaced values with `linspace`

`np.linspace(start, stop, num)` returns a chosen number of evenly spaced values. Unlike `arange`, the `stop` value is included by default.

```python
five_points = np.linspace(0, 10, 5)
percentages = np.linspace(0, 1, 11)

print("Five points from 0 to 10:", five_points)
print("Eleven points from 0 to 1:", percentages)
```

> **Key distinction:** Use `arange` when the step size matters. Use `linspace` when the number of points matters.

## 5. Reshaping arrays

`reshape` changes an array's dimensions without changing its data. The new shape must contain the same total number of elements.

```python
numbers = np.arange(1, 13)
reshaped_numbers = numbers.reshape(3, 4)

print("Original array:", numbers)
print("Reshaped as 3×4:\n", reshaped_numbers)
print("Flattened again:", reshaped_numbers.ravel())
```

## 6. Slices are views

A NumPy slice usually creates a **view** of the original array instead of an independent object. Changing the view therefore changes the original array too.

```python
original_array = np.arange(0, 15)
array_view = original_array[4:10]

print("Before the change:")
print("Original:", original_array)
print("View:", array_view)

array_view[:] = 100

print("\nAfter changing the view:")
print("Original:", original_array)
print("View:", array_view)
```

This differs from ordinary Python list slicing, which creates a new list.

```python
python_list = list(range(15))
list_slice = python_list[4:10]
list_slice[0] = -100

print("Original list:", python_list)
print("Modified list slice:", list_slice)
```

## 7. Creating independent arrays with `copy`

Use `.copy()` when changes must not affect the original NumPy array.

```python
original_array = np.arange(0, 20)
independent_copy = original_array.copy()
copied_slice = independent_copy[4:9]

copied_slice[:] = -100

print("Original array:", original_array)
print("Modified copy:", independent_copy)
print("Modified slice of the copy:", copied_slice)
```

> **Practical rule:** Slicing is memory-efficient because it usually returns a view. Use `.copy()` only when an independent array is required.

## 8. Boolean masking

A comparison applied to an array returns a boolean array. This mask can then be used to select only the values that satisfy the condition.

```python
random_values = rng.integers(1, 150, size=25)
greater_than_50 = random_values > 50

print("Random values:", random_values)
print("Boolean mask:", greater_than_50)
print("Values greater than 50:", random_values[greater_than_50])
```

Conditions can be combined with `&` (and), `|` (or), and `~` (not). Each condition must be enclosed in parentheses.

```python
between_50_and_100 = (random_values > 50) & (random_values < 100)
even_values = random_values[random_values % 2 == 0]

print("Values between 50 and 100:", random_values[between_50_and_100])
print("Even values:", even_values)
print("Number of values above 50:", np.count_nonzero(greater_than_50))
```

## 9. Vectorized arithmetic

Arithmetic between arrays of the same shape is performed element by element. Arithmetic with a scalar applies that value to every element.

```python
first_array = np.arange(1, 6)
second_array = np.array([10, 20, 30, 40, 50])

print("Addition:", first_array + second_array)
print("Subtraction:", second_array - first_array)
print("Multiplication:", first_array * second_array)
print("Division:", second_array / first_array)
print("Squared values:", first_array ** 2)
print("Add a scalar:", first_array + 100)
```

## 10. Universal functions and aggregations

NumPy's universal functions, often called **ufuncs**, apply mathematical operations element by element. Aggregations reduce many values to a summary value.

```python
values = np.arange(0, 6)

print("Square roots:", np.sqrt(values))
print("Exponential values:", np.exp(values))
print("Natural logarithms of values + 1:", np.log1p(values))
print("Maximum:", np.max(values))
print("Minimum:", np.min(values))
print("Mean:", np.mean(values))
print("Standard deviation:", np.std(values))
print("Sum:", np.sum(values))
```

Division by zero produces `nan` or `inf` instead of stopping the program with a Python exception. NumPy also emits a warning, so invalid values should be handled deliberately.

```python
numerators = np.array([0.0, 1.0, 2.0])
denominators = np.array([0.0, 0.0, 2.0])

with np.errstate(divide="ignore", invalid="ignore"):
    division_result = numerators / denominators

print("Division result:", division_result)
print("Is NaN:", np.isnan(division_result))
print("Is infinite:", np.isinf(division_result))
```

## 11. Mini practice

The following example combines array creation, reshaping, boolean masking, and aggregation.

```python
practice_matrix = rng.integers(1, 101, size=12).reshape(3, 4)
above_average = practice_matrix[practice_matrix > practice_matrix.mean()]

print("Practice matrix:\n", practice_matrix)
print("Shape:", practice_matrix.shape)
print("Mean:", practice_matrix.mean())
print("Values above the mean:", above_average)
print("Row sums:", practice_matrix.sum(axis=1))
print("Column sums:", practice_matrix.sum(axis=0))
```
